```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join(os.path.dirname("/medpop/esp2/lli/somalink_olink"), "..")))
project_root = os.path.abspath(os.path.expanduser("/medpop/esp2/lli/somalink_olink"))
sys.path.append(project_root)
os.chdir(project_root)

import warnings
import pickle
import pandas as pd
import numpy as np
import lightgbm as lgb

def fxn():
    warnings.warn("deprecated", DeprecationWarning)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    fxn()
```

```python
MESA_Olink  = pd.read_csv("Data/final_datasets/MESA_Olink_3k_log2_nodup_rm10na_knn_final.tsv.gz", sep = "\t",
                          compression = "gzip")
MESA_Soma = pd.read_csv("Data/final_datasets/MESA_SOMA_5k_log2scaled_nodup_final.tsv.gz", sep = "\t",
                        compression = "gzip").drop(columns={"bmi"}, axis=1)
```

```python
# common_cols = ["id"] + [col for col in MESA_Soma.columns if col in MESA_Olink.columns and col != "id"]
# MESA_Soma  = MESA_Soma[common_cols]
# MESA_Olink = MESA_Olink[common_cols]
```

```python
overlap_IDs = np.unique(list(set(np.array(MESA_Olink["id"])).intersection(set(np.array(MESA_Soma["id"])))))
```

```python
MESA_Olink  = MESA_Olink[MESA_Olink["id"].isin(overlap_IDs)].sort_values(by=["id"]).drop_duplicates(["id"])
MESA_Soma   = MESA_Soma[MESA_Soma["id"].isin(overlap_IDs)].sort_values(by=["id"]).drop_duplicates(["id"])

print(MESA_Olink.shape)
print(MESA_Soma.shape)

ids1 = list(MESA_Olink["id"])
ids2 = list(MESA_Soma["id"])

if ids1 == ids2:
    print("ids in the same order")
else:
    print("ids not in the same order")
```

```python
MESA_Soma = MESA_Soma.set_index("id")
MESA_Olink = MESA_Olink.set_index("id")

if MESA_Soma.index.equals(MESA_Olink.index):
    print("yes")
else:
    print("no")
```

```python
## SAME pickled splits used for HGB, so results are directly comparable
train_index = pickle.load(open(f"Data/helper_data/train_index.pkl", "rb"))
test_index = pickle.load(open(f"Data/helper_data/test_index.pkl", "rb"))

print(train_index[1:20])
```

```python
## SAME chunking scheme as HGB batches -- note this uses the SO protlist (Olink target names),
## adjust the pickle filename if your SO/OS chunks are stored under different names
protlist = pickle.load(open(f"Data/helper_data/so_protlist_chunk{batch_number}.pkl", "rb"))
print(protlist[0:10])
print(len(protlist))
```

## Training loop (LightGBM, SomaScan -> Olink)

```python
## defining X -- SomaScan is the feature set in this direction
X = MESA_Soma
print(MESA_Soma.shape)
```

```python
## default hyperparameters (no tuning), matching the same "reasonable defaults" framing
## used for HGB, so this is an apples-to-apples benchmark
def train_soma_to_olink_lgb(protein_name):
    y = MESA_Olink[protein_name]
    if y.loc[train_index].isna().any():
        print(f"skip protein {protein_name} due to NA in OLINK data")
        return

    X_train = X.loc[train_index]
    y_train = y.loc[train_index]

    reg = lgb.LGBMRegressor(random_state=0, verbosity=-1)
    reg.fit(X_train, y_train)

    model_path = f"Data/soma_trained_models/soma_olink/5k_log2scaled_soma_to_3kolink_lgb/{protein_name}_lgb.pkl"
    os.makedirs(os.path.dirname(model_path), exist_ok=True)
    pickle.dump(reg, open(model_path, "wb"))
    print(f"Saved model for {protein_name} to {model_path}")
```

```python
%%time

for p in protlist:
    train_soma_to_olink_lgb(p)
```
